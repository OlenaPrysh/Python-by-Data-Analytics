3.1. Створіть зводну таблицю, яка покаже, скільки покупок (інвойсів) було зроблено в кожній з філій (`Branch`) за різними категоріями товарів. Запишіть таблицю в змінну `invoices_by_category` і виведіть змінну на екран.

3. Які категорії товарів формують найбільший прибуток по філіалах. Це допоможе зрозуміти локальний попит. Допомагає оптимізувати асортимент товарів.

```python
df['Date'] = pd.to_datetime(df['Date'])
df['weekday'] = df['Date'].dt.day_name()
df.groupby('weekday')['Total'].agg(['count', 'mean']).sort_values(by='count', ascending=False)
```

```python
weekend_df = df[df['weekday'].isin(['Saturday', 'Sunday'])]
weekend_df['Product line'].value_counts(normalize=True)
```

```python
profit_by_category = df.groupby(['Branch', 'Product line'])['gross income'].sum().reset_index()
top3_profit = profit_by_category.sort_values(['Branch', 'gross income'], ascending=[True, False]).groupby('Branch').head(3)

print(top3_profit)
```
